Root Cause Found
The low intra-class similarity is due to pattern sparsity + peak position shifts:

Issue	Detail
Pattern sparsity	Only 0.11% of points are non-zero (sharp peaks)
Peak shifts	Same compound, slightly different cell parameters → peaks shift by a few indices
Cosine similarity	Sharp peaks at different positions = near-zero dot product

Conclusion
Good news: Most classes are actually consistent - the same compound produces similar XRD patterns with minor peak shifts due to:

Different temperature/pressure conditions
Slight compositional variations
Different CIF sources
True polymorphs (still low similarity after smoothing):

Neon, Hydrogen, Helium, Oxygen - gases with multiple solid phases
Steinhardtite, Pyrolusite - known polymorphic minerals
Implication for Classification
The classifier should be robust to small peak shifts. Options:

Use smoothed patterns for training
Use DTW-based similarity instead of cosine
Add shift-invariant augmentation during training
